```python
#Function to write data into text file of climate,technology and sports 
def write_data(url,class_name):   
        page = requests.get(url)
        soup= BeautifulSoup(page.text,"html.parser")  #crawling the RSS feed url
        data = soup.find_all('p',class_=class_name)   #getting the data from page
        line=0
        linecount = 0
        for datadict in data:
            file = open ('C:\Documents\Bharathi ed\IR\ClusterData.txt','a',encoding='UTF-8')  # writing the data into the file
            file.write(datadict.text)
            line+=1
            if(class_name != 'qa-introduction gel-pica-bold'):    # for Climate and technology getting the data of this class
                if (line==3):
                    file.write('\n')
                    linecount += 1 
                    line=0
                    if (linecount == 2):
                        break;
            else:                                  # for Sports data 
                file.write(entry.title)             
                file.write(entry.description)
                if (line==1):
                    file.write('\n')
                    linecount += 1 
                    line=0
                    if (linecount == 2):
                        break;
```

```python
#opening the file in write mode
file = open ('C:\Documents\Bharathi ed\IR\ClusterData.txt','w',encoding='UTF-8')
#writing the data of climate into file
for entry in climate.entries:
    climate_url = entry.link
    climate_class_name = 'ssrcss-1q0x1qg-Paragraph eq5iqo00'
    write_data(climate_url,climate_class_name)         
```

```python
#writing the data of technology into file
for entry in technology.entries:
    technology_url = entry.link
    technology_class_name = 'ssrcss-1q0x1qg-Paragraph eq5iqo00'
    write_data(technology_url,technology_class_name)           
```

```python
#writing the data of Sports into file
for entry in sports.entries:
    sports_url = entry.link
    sports_class_name = 'qa-introduction gel-pica-bold'
    write_data(sports_url,sports_class_name) 
file.close() # closing file
```

```python
#opening the file in read mode  and reading the data
file = open ('C:\Documents\Bharathi ed\IR\ClusterData.txt','r',encoding='UTF-8')
documents = file.read().split("\n")
documents
```

```python
#importing nltk and download stop words
import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords
stop_word = stopwords.words('english')
print(stop_word)
```

```python
# libraries
nltk.download("punkt")
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
Porter_Stem = PorterStemmer()
```

```python
# function for tokenize and stemmimg
def token_stemmer(documents):
    filtered_documents = []
    for doc in documents:
        tokens = word_tokenize(doc)
        temp = ""
        for word in tokens:
            if word not in stop_word:
                temp += Porter_Stem.stem(word) + " "
        filtered_documents.append(temp)
    return(filtered_documents)
```

```python
#stemming and tokenize the document
filtered_documents = token_stemmer(documents)
filtered_documents
```

```python
#vectorize
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(filtered_documents)
print(X.todense())
```

```python
#implementing K-means clustering
from sklearn.cluster import KMeans
K=3
model = KMeans(n_clusters=K,random_state=11)
model.fit(X)
labels = model.labels_
print("Clusters of Documents in the file:\n",model.labels_)
```

```python
#Predicting the model
Z = vectorizer.transform(filtered_documents)
prediction = model.predict(Z)
print(prediction)
```

```python
#test data file
testfile = open ('Testdata.txt','r')
test_document = testfile.read().split(";")
test_document
```

```python
#stemming and tokenize the test data
filtered_testdoc = token_stemmer(test_document)
filtered_testdoc
```

```python
#predicting that test data is allocated to right cluster
testdata = vectorizer.transform(filtered_testdoc)
prediction_testdata = model.predict(testdata)
print(prediction_testdata)
```

```python
# silhouette Scoring for K means
from sklearn.metrics import silhouette_score
silhouette_score_kmeans = sklearn.metrics.silhouette_score(X,labels)
print("silhouette score of kmeans Clustering Algorithm:",silhouette_score_kmeans*100)
```

```python
# calinski harabasz Scoring for K means
from sklearn.metrics import calinski_harabasz_score
calinskiharabasz_score = sklearn.metrics.calinski_harabasz_score(X.toarray(),labels)
print("calinski harabasz score of kmeans Clustering Algorithm:",calinskiharabasz_score*100)
```
