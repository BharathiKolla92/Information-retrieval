```python
import requests
from bs4 import BeautifulSoup
```

```python
def htmlparser(page_url):
    page = requests.get(page_url)
    soup= BeautifulSoup(page.text,"html.parser")
    return soup
```

```python
url ="https://pureportal.coventry.ac.uk/en/organisations/research-centre-for-computational-science-and-mathematical-modell"
main_page_soup = htmlparser(url)
main_page_soup
```

```python
main_page_soup.find_all('a',class_='portal_link')
```

```python
def subpage_url(href_sub):
    subpage = main_page_soup.find('a',href = href_sub)
    subpage_link = subpage.get('href')
    subpage_fulllink = url+subpage_link[3:]
    return subpage_fulllink
```

```python
member_profile_fulllink = subpage_url("/en/persons/")
member_profile_fulllink
```

```python
full_publications_link = subpage_url("/en/publications/")
full_publications_link
```

```python
member_profile_soup=htmlparser(member_profile_fulllink)
member_profile_soup
```

```python
csm_member_profile = member_profile_soup.find_all('a',class_="link person")
csm_member_profile_name = []
for member in csm_member_profile:
    csm_member_profile_name.append(member['href'])
print("CSM Members are :\n",csm_member_profile_name[0])
```

```python
publications_page_soup = htmlparser(full_publications_link)
publications_page_soup
```

```python
first_publication_page = full_publications_link 
next_publication_pages=publications_page_soup.find_all('a',class_='step')
all_publication_pages_url = []
all_publication_pages_url.append(first_publication_page)
for page in next_publication_pages:
    next_publication_page_url = page.get('href')
    page_url = full_publications_link + next_publication_page_url[-7:]
    all_publication_pages_url.append(page_url)
all_publication_pages_url
```

```python
Publication_name = []
Publication_Link = []
Author_name_pub = []
Author_Link_pub = []
Publication_date = []
```

```python
for publication_page in all_publication_pages_url:
    publications_page_soup= htmlparser(publication_page)
    for link in publications_page_soup.find_all('div',class_='result-container'):
            Publication_name.append(link.span.text)
            Publication_Link.append(link.a['href'])
            date = link.find('span',class_='date')
            Author_name = []
            span_list = link.find_all('span')
            date = link.find('span',class_='date')
            translated_title = link.find('span',class_='translated_title')
            for author in span_list:
                if(author.text != date.text):
                    if(author.text != link.span.text):
                        if(translated_title == None):
                            Author_name.append(author.text)   
                        else:
                            if(author.text != translated_title.text):
                                Author_name.append(author.text) 
                else:
                    break;
            Author_Link = []
            for person in link.find_all('a',class_='link person'):
                Author_Link.append(person.attrs['href'])
            Author_name_pub.append(Author_name)
            Author_Link_pub.append(Author_Link)
            date = link.find('span',class_='date')
            Publication_date.append(date.text)
```

```python
for i in range(len(Publication_name)):
    print(Publication_name[i])
    print(Publication_Link[i])
    print(Author_name_pub[i])
    for j in range(len(Author_Link_pub[i])):
        print(Author_Link_pub[i][j])
    print(Publication_date[i])
    
    print("****************************************************************************************************************")
```

```python
import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords
sw = stopwords.words('english')
additionalsw = ['[','.',']',',', "'",'https',':','``']
sw.extend(additionalsw)
print(sw)
```

```python
nltk.download("punkt")
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
ps = PorterStemmer()
filtered_docs = []
for doc in docs:
    tokens = word_tokenize(doc)
    tmp = ""
    for w in tokens:
        if w not in sw:
            tmp = ps.stem(w)
            filtered_docs.append(tmp)
    
print(filtered_docs)
```

Inverted Index

```python
inverted_index = {}
document_number = 1
for doc in docs:
    tokens = word_tokenize(doc)
    tmp = ""
    for w in tokens:
        if w not in sw:
            tmp = ps.stem(w)+ " "
            if tmp not in inverted_index:
                inverted_index[tmp]=[]
            inverted_index[tmp].append(document_number)
    document_number +=1
print(inverted_index)
```
